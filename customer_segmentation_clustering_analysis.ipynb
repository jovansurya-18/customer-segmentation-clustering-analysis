# Data Cleansing and Preprocessing

```python
df = pd.read_csv('https://github.com/FourthRevGitHub/CAM_DS_Public-Data-Sets/raw/main/Course%201/Week%206/CUSTOMERS_CLEAN.zip')
df
```

```python
df.info()
```

```python
df.describe()
```

```python
#check for nulls

df.isnull().sum()
```

```python
# check the % of the nulls

(df.isnull().sum()/len(df) * 100).round(2)

# the missing values for state_province is concerning, and should actually be investigated in a rela life situtation
# although it is definsible to also drop the null samples for "City" and "Postal_Code", i would opt to replacing the nulls with "Unknown", so lets do that
```

```python
# replacing nulls with Unknown
df["City"] = df["City"].fillna("Unknown")
df["Postal_Code"] = df["Postal_Code"].fillna("Unknown")
df["State_Province"] = df["State_Province"].fillna("Unknown")

df.isnull().sum()
```

```python
# assigning the datetime dtype to date columns

date_columns = ["Order_Date", "Delivery_Date", "Customer_BirthDate"]
df[date_columns] = df[date_columns].apply(pd.to_datetime)
df
```

```python
# clean and assign float dtype to currency columns

curr_columns = ["Profit", "Total Revenue", "Unit Cost"]

for curr_column in curr_columns:
  df[curr_column] = df[curr_column].astype(str)
  df[curr_column] = df[curr_column] \
    .str.replace("$", "") \
    .str.replace("(", "-") \
    .str.replace(")", "") \
    .str.replace(",", "") \
    .astype(float)
  df.rename(columns={curr_column: f"{curr_column} USD"}, inplace = True)


# change "Customer ID" and "Order ID" to str as we're not performing any arithmethic on it

df["Customer ID"] = df["Customer ID"].astype(str)
df["Order ID"] = df["Order ID"].astype(str)


# clean and assign float dtype to "Discount"

df["Discount"] = df["Discount"].astype(str)
df["Discount"] = df["Discount"] \
  .str.strip() \
  .str.replace("%", "") \
  .str.replace(".", "0") \
  .astype(float)
df.rename(columns={"Discount": "Discount_Percentage"}, inplace=True)


df.head()
```

```python
# check dtypes
df.info()
```

```python
# check for duplicates

df.duplicated().sum()
```

```python
# lets remove those duplicates

df.drop_duplicates(inplace = True)
df.duplicated().sum()
```

```python
# lets change all columns name to snake case for consistency

df.rename(columns={
    "Total Revenue USD": "Total_Revenue_USD",
    "Unit Cost USD": "Unit_Cost_USD",
    "OrderTypeLabel": "Order_Type_Label",
    "CustomerCountryLabel": "Customer_Country_Label",
    "Order ID": "Order_ID",
    "Profit USD": "Profit_USD",   # fixed typo "Profit USE" → "Profit USD"
    "Days to Delivery": "Days_to_Delivery",
    "Loyalty Num": "Loyalty_Num",
    "Customer ID": "Customer_ID"
}, inplace=True)

df.head()
```

```python
for column in df.columns:
  print(f"Number of unique values in {column}: {df[column].nunique()}")
```

```python
# it seems we won't do any arithmetics with "Loyalty_Num"

df["Loyalty_Num"] = df["Loyalty_Num"].astype(str)
```

# Feature Transformation

```python
# create new features for frequency, recency, CLV, average unit cost, and customer age
# i will create a new df with consisting all unique customer id with all the extra new features

# frequency
df_freq = (
    df.groupby("Customer_ID")["Order_ID"]
    .nunique()
    .reset_index()
    .rename(columns={"Order_ID": "Frequency"})
)


# recency
# we'll use the latest order in the df as the cut-off date to calculate the recency of each order
cutoff_date = df["Order_Date"].max()

df_recency = (
    df.groupby("Customer_ID")["Order_Date"]
    .max()
    .reset_index()
    .rename(columns={"Order_Date": "Last_Order_Date"})
)

df_recency["Recency_in_Days"] = (cutoff_date - df_recency["Last_Order_Date"]).dt.days
df_recency.drop(columns=["Last_Order_Date"], inplace=True)


# CLV
df_clv = (
    df.groupby("Customer_ID")["Profit_USD"]
    .sum()
    .reset_index()
    .rename(columns={"Profit_USD": "CLV_USD"})
)


# avg unit cost
df_auc = (
    df.groupby("Customer_ID")["Unit_Cost_USD"]
    .mean()
    .reset_index()
    .rename(columns={"Unit_Cost_USD": "Avg_Unit_Cost_USD"})
)


# customer age
# cut-off date will be today
date_today = pd.to_datetime("today").normalize()
df_age = (
    df.groupby("Customer_ID")["Customer_BirthDate"]
    .max()
    .reset_index()
    .rename(columns={"Customer_BirthDate": "Birthdate"})
)

df_age["Age_in_Years"] = (date_today - df_age["Birthdate"]).dt.days//365
df_age.drop(columns=["Birthdate"], inplace=True)
```

```python
# merge all of the new features in a new df

df_cust = df_freq
df_cust = df_cust.merge(df_recency, on = "Customer_ID", how = "left")
df_cust = df_cust.merge(df_clv, on = "Customer_ID", how = "left")
df_cust = df_cust.merge(df_auc, on = "Customer_ID", how = "left")
df_cust = df_cust.merge(df_age, on = "Customer_ID", how = "left")
df_cust
```

# Determining k for clustering

Elbow method

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaled_df = scaler.fit_transform(df_cust)
```

```python
from sklearn.cluster import KMeans

wcss = []

for i in range(1,11):
  kmeans = KMeans(
      n_clusters = i,
      init = "k-means++",
      random_state = 42,
      n_init = 10
      )
  kmeans.fit(scaled_df)
  wcss.append(kmeans.inertia_)

plt.plot(
    range(1,11),
    wcss,
    marker = "o"
)
plt.title("Elbow Method")
plt.xlabel("Number of Clusters")
plt.ylabel("WCSS")
plt.show()

# the elbow method seems to suggest that k=6 is optimal number of clusters
# but the chart is also relatively vague, lets try using the silhouette method
```

Silhouette method

```python
from sklearn.metrics import silhouette_score

for i in range (2,9):
  kmeans = KMeans(
      n_clusters = i,
      init = "k-means++",
      random_state = 42,
      n_init = 10
  )
  label = kmeans.fit_predict(scaled_df)
  silhouette_avg = silhouette_score(scaled_df, label)
  print(f"The Silhouette score for k = {i} is {silhouette_avg}\n")

# the Silhouette score shows that k=5 is the most optimal number of clusters
```

Agglomerative Hierarchical Clustering

```python
from scipy.spatial.distance import pdist, squareform
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster

columns = df_cust.columns
scaled_df = pd.DataFrame(scaled_df, columns = columns)
sample_df = scaled_df.sample(n=100, random_state = 42)

dists = pdist(sample_df, metric = "euclidean")
linkage_matrix = linkage(dists, method = "average")

plt.figure(figsize=(20,10))
dendrogram(linkage_matrix)
plt.title("Dendrogram with Average Linkage")
plt.xlabel("Customers")
plt.ylabel("Distance")
plt.show()
```

```python
linkage_matrix = linkage(dists, method = "ward")

plt.figure(figsize=(20,10))
dendrogram(linkage_matrix)
plt.title("Dendrogram with Ward Linkage")
plt.xlabel("Customers")
plt.ylabel("Distance")
plt.show()
```

```python
linkage_matrix = linkage(dists, method = "complete")

plt.figure(figsize=(20,10))
dendrogram(linkage_matrix)
plt.title("Dendrogram with Complete Linkage")
plt.xlabel("Customers")
plt.ylabel("Distance")
plt.show()
```

```python
linkage_matrix = linkage(dists, method = "single")

plt.figure(figsize=(20,10))
dendrogram(linkage_matrix)
plt.title("Dendrogram with Single Linkage")
plt.xlabel("Customers")
plt.ylabel("Distance")
plt.show()
```

```python
linkage_matrix = linkage(dists, method = "centroid")

plt.figure(figsize=(20,10))
dendrogram(linkage_matrix)
plt.title("Dendrogram with Centroid Linkage")
plt.xlabel("Customers")
plt.ylabel("Distance")
plt.show()
```

# K-mean Clustering

```python
kmeans = KMeans(
    n_clusters = 5,
    init = "k-means++",
    random_state = 42,
    n_init = 10
)

scaled_df = scaled_df.drop(columns=["Customer_ID"])

scaled_df["Cluster"] = kmeans.fit_predict(scaled_df)
scaled_df
```

```python
columns = scaled_df.columns[:-1]

plt.figure(figsize=(20,10))
for i, column in enumerate(columns):
  plt.subplot(2,3,i+1)
  sns.boxplot(
      data = scaled_df,
      x = scaled_df["Cluster"],
      y = scaled_df[column]
  )
  plt.title(column)

plt.tight_layout()
plt.show()
```

```python
# dimensionality reduction

# PCA

from sklearn.decomposition import PCA

scaled_df_cluster = scaled_df.drop(columns=["Cluster"])

pca = PCA(n_components = 2)
X_pca = pca.fit_transform(scaled_df_cluster)
X_pca.shape
```

```python
plt.figure(figsize=(10,10))
sns.scatterplot(
    x = X_pca[:,0],
    y = X_pca[:,1],
    hue = scaled_df["Cluster"]
)

plt.title("PCA 2D")
plt.xlabel("PCA1")
plt.ylabel("PCA2")
plt.show()
```

```python
# t-SNE

from sklearn.manifold import TSNE

sample_df_cluster = scaled_df.sample(n = 1000, random_state = 42) # to reduce processing time
tsne = TSNE(n_components = 2, perplexity = 30.0)
X_tsne = tsne.fit_transform(sample_df_cluster.drop(columns=["Cluster"]))
X_tsne.shape
```

```python
sns.scatterplot(
    x = X_tsne[:, 0],
    y = X_tsne[:, 1],
    hue = sample_df_cluster["Cluster"]
)

plt.title("t-SNE 2D")
plt.xlabel("t-SNE_1")
plt.ylabel("t-SNE_2")
plt.show()

# t-sne performs way better for visualisation, especially when its compared to PCA
```
